```python
from transformers import RobertaModel, RobertaTokenizer
from bertviz import model_view, head_view
```

```python
# model_version = 'seyonec/PubChem10M_SMILES_BPE_396_250'
model_name = "DeepChem/ChemBERTa-77M-MTR"
model = RobertaModel.from_pretrained(model_name, output_attentions=True)
tokenizer = RobertaTokenizer.from_pretrained(model_name)
```

## TOKENIZER


```python
# smiles = "CN1C=NC2=C1C(=O)N(C(=O)N2C)C" # Caffeine
# smiles = "C1=CC=CC=C1"  # Benzene
smiles = "CN=C=O"

# smiles = "CNC1CC(c2ccc(Cl)c(Cl)c2)c2ccccc21"

# basic_tokens = tokenizer.basic_tokenizer.tokenize(smiles)
tokens = tokenizer.tokenize(smiles)

# print(basic_tokens)
print(tokens)
```

```python
# Get tokenizer output
inputs = tokenizer(smiles, return_tensors="pt", add_special_tokens=True)
input_ids = inputs["input_ids"]
# attention = model(input_ids)[-1]
input_id_list = input_ids[0].tolist()
tokens = tokenizer.convert_ids_to_tokens(input_id_list)
tokens
```

```python
sentence_a = "CCCCC[C@@H](Br)CC"
sentence_b = "CCCCC[C@H](Br)CC"

inputs = tokenizer.encode_plus(
    sentence_a, sentence_b, return_tensors="pt", add_special_tokens=True
)
input_ids = inputs["input_ids"]
attention = model(input_ids)[-1]
input_id_list = input_ids[0].tolist()  # Batch index 0
tokens = tokenizer.convert_ids_to_tokens(input_id_list)
model_view(attention, tokens)
```
