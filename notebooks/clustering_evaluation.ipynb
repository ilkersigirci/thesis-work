```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import homogeneity_score, adjusted_rand_score

# Generate a random dataset with 3 clusters and 100 samples per cluster
X, y = make_blobs(n_samples=300, centers=3, random_state=0)

# Apply KMeans clustering with 3 clusters
kmeans = KMeans(n_clusters=3, random_state=0)
labels = kmeans.fit_predict(X)

# Calculate the homogeneity score and adjusted rand score
homogeneity = homogeneity_score(y, labels)
ari = adjusted_rand_score(y, labels)

# Print the scores
print("Homogeneity score:", homogeneity)
print("Adjusted Rand Index:", ari)
```

# homogeneity_score

```python
# Perfect labelings are homogeneous:

from sklearn.metrics.cluster import homogeneity_score

homogeneity_score([0, 0, 1, 1], [1, 1, 0, 0])
```

```python
# Non-perfect labelings that further split classes into more clusters can be perfectly homogeneous:
print("%.6f" % homogeneity_score([0, 0, 1, 1], [0, 0, 1, 2]))
print("%.6f" % homogeneity_score([0, 0, 1, 1], [0, 1, 2, 3]))
```

```python
# Clusters that include samples from different classes do not make for an homogeneous labeling:

print("%.6f" % homogeneity_score([0, 0, 1, 1], [0, 1, 0, 1]))
print("%.6f" % homogeneity_score([0, 0, 1, 1], [0, 0, 0, 0]))
```

# adjusted_rand_score

```python
# Perfectly matching labelings have a score of 1 even

from sklearn.metrics.cluster import adjusted_rand_score

print(adjusted_rand_score([0, 0, 1, 1], [0, 0, 1, 1]))
print(adjusted_rand_score([0, 0, 1, 1], [1, 1, 0, 0]))
```

```python
# Labelings that assign all classes members to the same clusters are complete but may not always be pure, hence penalized:
adjusted_rand_score([0, 0, 1, 2], [0, 0, 1, 1])
```

```python
# ARI is symmetric, so labelings that have pure clusters with members coming from the same classes but unnecessary splits are penalized:  (IMPORTANT !!!)

adjusted_rand_score([0, 0, 1, 1], [0, 0, 1, 2])
```

```python
# If classes members are completely split across different clusters, the assignment is totally incomplete, hence the ARI is very low:

adjusted_rand_score([0, 0, 0, 0], [0, 1, 2, 3])
```

```python
# ARI may take a negative value for especially discordant labelings that are a worse choice than the expected value of random labels:
adjusted_rand_score([0, 0, 1, 1], [0, 1, 0, 1])
```
